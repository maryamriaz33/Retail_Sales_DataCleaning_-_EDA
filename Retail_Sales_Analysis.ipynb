```python
df["transaction_date"] = pd.to_datetime( df["transaction_date"])
```

```python
df['month'] = df['transaction_date'].dt.month
```

```python
df['year'] = df['transaction_date'].dt.year
```

```python
# Feature Engineering
df['total_sales'] = df['total_spent']
```

```python
df['day_of_week'] = df['transaction_date'].dt.day_name()
```

```python
df['average_order_value'] = df['total_sales'].mean()
```

```python
df['customer_total_spending'] = df.groupby('customer_id')['total_sales'].transform('sum')
```
